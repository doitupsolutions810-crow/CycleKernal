```
#!/usr/bin/env python3
import os
import sys
import time
import requests
import json
from datetime import datetime

# --- CONFIGURATION ---
API_BASE = "http://localhost:9002/api"
SYSTEM_NAME = "AXIOM EVOLUTION TERMINAL"
VERSION = "v6.0.4-STABLE"

class AxiomTerminal:
    def __init__(self):
        self.active = True
        self.session_start = datetime.now()

    def clear(self):
        os.system('cls' if os.name == 'nt' else 'clear')

    def header(self):
        print(f"\033[1;32m{'='*60}\033[0m")
        print(f"\033[1;37m {SYSTEM_NAME} | {VERSION}\033[0m")
        print(f"\033[1;30m KERNEL LINK: {API_BASE}\033[0m")
        print(f"\033[1;32m{'='*60}\033[0m\n")

    def fetch_status(self):
        try:
            response = requests.get(f"{API_BASE}/wallet/status", timeout=2)
            if response.status_code == 200:
                return response.json()
        except:
            return None

    def toggle_mining(self):
        try:
            response = requests.post(f"{API_BASE}/mining/toggle", timeout=2)
            if response.status_code == 200:
                return response.json()
        except:
            return None

    def display_dashboard(self, data):
        if not data:
            print("\033[1;31m[!] KERNEL OFFLINE: Target localhost:9002 unreachable.\033[0m")
            print("    Execute 'python3 launch_system.py' to establish link.\n")
            return

        res_status = "\033[1;32mENGAGED (ABSORBING SHARDS)\033[0m" if data['resonance_active'] else "\033[1;30mIDLE\033[0m"

        print(f" \033[1;37mNODE UID:\033[0m {data['wallet_id']}")
        print(f" \033[1;37mRESONANCE:\033[0m {res_status}")
        print(f" {'-'*30}")
        print(f" \033[1;32mBTC BALANCE:\033[0m {data['available_balance']:.8f} BTC")
        print(f" \033[1;32mUSD VALUE:\033[0m   ${data['usd_value']:,.2f}")
        print(f" \033[1;32mPOOL RECAP:\033[0m  {data['pooled_balance']:.8f} BTC")
        print(f" {'-'*30}")
        print(f" \033[1;37mESTABLISHED EXCHANGE:\033[0m 1 BTC = ${data['exchange_rate']['bitcoin']:,.2f}")
        print(f"\n \033[1;30mLAST SYNC: {datetime.fromtimestamp(data['timestamp']).strftime('%H:%M:%S')}\033[0m\n")

    def run(self):
        while self.active:
            self.clear()
            self.header()
            data = self.fetch_status()
            self.display_dashboard(data)

            print("\033[1;37m[1]\033[0m Toggle Shard Resonate (Absorption)")
            print("\033[1;37m[2]\033[0m Refresh Telemetry")
            print("\033[1;37m[3]\033[0m Audit Kernel Logs")
            print("\033[1;31m[Q]\033[0m Disconnect Terminal")

            choice = input("\n\033[1;32mAXIOM_CMD >> \033[0m").upper()

            if choice == '1':
                result = self.toggle_mining()
                if result:
                    print(f"\n[✓] Command broadcasted: Resonance {result['mining']}.")
                    time.sleep(1)
            elif choice == '2':
                continue
            elif choice == '3':
                print("\n[!] Accessing Kernel Log Stream...")
                time.sleep(1)
                print("    > Handshake verified.")
                print("    > Port 9002 bound to local interface.")
                input("\nPress Enter to return...")
            elif choice == 'Q':
                print("\n[!] Disconnecting...")
                self.active = False
            else:
                print("\n[!] Invalid Command.")
                time.sleep(0.5)

if __name__ == "__main__":
    try:
        terminal = AxiomTerminal()
        terminal.run()
    except KeyboardInterrupt:
        print("\n\n[!] Terminal process killed.")
        sys.exit(0)
```
